# EXP3.1 — Parallel Runner

Executes all three EXP3.1 notebooks (A, B, C) in sequence using `nbconvert`.

| Notebook | Models | Search Budget |
|---|---|---|
| **EXP3.1-A** | KNN + Random Forest | 30 S1 trials, 6 S2 refine, 600 final epochs |
| **EXP3.1-B** | XGBoost + AdaBoost | 40 S1 trials, 6 S2 refine, 600 final epochs |
| **EXP3.1-C** | Logistic Regression + CatBoost | 30 S1 trials, 6 S2 refine, 600 final epochs |

**Results saved to:** `exp3_1_knn_rf/`, `exp3_1_xgb_ada/`, `exp3_1_logreg_cat/`

> Run cells one at a time to train each sub-experiment, or run all at once.  
> Each cell executes its notebook in-place via `nbconvert --execute`.

```python
import subprocess
import sys
import time
from pathlib import Path

NOTEBOOKS = [
    'EXP3_1_A_KNN_RF.ipynb',
    'EXP3_1_B_XGB_AdaBoost.ipynb',
    'EXP3_1_C_LogReg_CatBoost.ipynb',
]

ROOT = Path.cwd()

def run_notebook(nb_name, timeout_seconds=14400):
    """Execute a notebook via nbconvert and report timing."""
    nb_path = ROOT / nb_name
    if not nb_path.exists():
        print(f'[ERROR] Not found: {nb_path}')
        return False

    print(f'\n{"="*70}')
    print(f'  Running: {nb_name}')
    print(f'{"="*70}')
    t0 = time.time()

    cmd = [
        sys.executable, '-m', 'jupyter', 'nbconvert',
        '--to', 'notebook',
        '--execute',
        '--inplace',
        f'--ExecutePreprocessor.timeout={timeout_seconds}',
        '--ExecutePreprocessor.kernel_name=python3',
        str(nb_path),
    ]

    result = subprocess.run(cmd, capture_output=False, text=True)

    elapsed = time.time() - t0
    status  = 'SUCCESS' if result.returncode == 0 else 'FAILED'
    print(f'\n  {status} — {nb_name}  ({elapsed/60:.1f} min)')
    return result.returncode == 0

print('EXP3.1 Runner ready.')
print('Notebooks to execute:', NOTEBOOKS)
print(f'Working directory: {ROOT}')
```

## EXP3.1-A — KNN + Random Forest
Runs the improved KNN + RandomForest experiment with tighter hyperparameter ranges.

```python
run_notebook('EXP3_1_A_KNN_RF.ipynb')
```

## EXP3.1-B — XGBoost + AdaBoost
Runs the improved XGBoost + AdaBoost experiment with broader regularisation search.

```python
run_notebook('EXP3_1_B_XGB_AdaBoost.ipynb')
```

## EXP3.1-C — Logistic Regression + CatBoost
Runs the improved LogReg + CatBoost experiment with a finer learning-rate lower bound.

```python
run_notebook('EXP3_1_C_LogReg_CatBoost.ipynb')
```

---
## Run ALL Three Sequentially
Alternatively, run this single cell to execute A → B → C in sequence.

```python
results = {}
for nb in NOTEBOOKS:
    results[nb] = run_notebook(nb)

print('\n' + '='*70)
print('EXP3.1 RUNNER — FINAL SUMMARY')
print('='*70)
for nb, ok in results.items():
    status = '✓ SUCCESS' if ok else '✗ FAILED'
    print(f'  {status}  {nb}')
print()
print('Results saved to:')
for d in ['exp3_1_knn_rf', 'exp3_1_xgb_ada', 'exp3_1_logreg_cat']:
    print(f'  {d}/')
```
